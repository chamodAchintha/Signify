## Graphs

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

def draw_plots(path):
    with open(path, "r") as f:
        data = f.read().split('\n')[:-1]
        
    train_losses = [float(d.split(' ')[4]) for d in data]
    validation_losses = [float(d.split(' ')[7][:-1]) for d in data]
    accuracies = [float(d.split(' ')[9]) for d in data]
    f1_scores = [float(d.split(' ')[11]) for d in data]
    epoches = [i+1 for i in range(len(train_losses))]
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # Plot train and validation losses
    ax1.plot(epoches, train_losses, label="Train Loss", color="blue")
    ax1.plot(epoches, validation_losses, label="Validation Loss", color="orange")
    ax1.set_xlabel("Epochs")
    ax1.set_ylabel("Loss")
    ax1.set_title("Train and Validation Loss")
    ax1.legend()
    
    # Plot accuracy and F1 score
    ax2.plot(epoches, accuracies, label="Accuracy", color="green")
    ax2.plot(epoches, f1_scores, label="F1 Score", color="purple")
    ax2.set_xlabel("Epochs")
    ax2.set_ylabel("Score")
    ax2.set_title("Validation Accuracy and F1 Score")
    ax2.legend()
    
    plt.tight_layout()
    plt.savefig(f"{'/'.join(path.split('/')[:-1])}/{path.split('/')[-2]}_loss.png")
    plt.show()
    


def plot_confusion_matrix(path):
    df = pd.read_csv(path)
    y_true = df.iloc[:, 0] 
    y_pred = df.iloc[:, 1] 
    
    cm = confusion_matrix(y_true, y_pred)
    
    plt.figure(figsize=(10, 10))
    sns.heatmap(cm, annot=False, fmt="d", cmap="viridis")
    plt.xlabel("Predicted Class")
    plt.ylabel("True Class")
    plt.title("Confusion Matrix for Multiclass Classification")
    plt.savefig(f"{'/'.join(path.split('/')[:-1])}/{path.split('/')[-2]}_accuracy_f1.png")
    plt.show()
```
